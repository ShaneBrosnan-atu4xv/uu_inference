## Problem 1

Computing the estiamted standard error:
$$
\widehat{SE}(\bar{X}) = \frac{s}{\sqrt{n}} = \frac{10}{\sqrt{100}} = \frac{10}{10} = 1
$$


Testing $H_0:\mu=50$ vs. $H_1:\mu\ne50$ to get the z-value and then p-value:
$$
z = \frac{\bar{X}-\mu_0}{\widehat{SE}(\bar{X})} = \frac{52-50}{1} = 2.00, \qquad p = 2(1-\Phi(2.00)) \approx 0.0455
$$


Interpreting the p-value at different significance levels:

| $\alpha$ | Critical value | Comparison | Decision |
|:---:|:---:|:---:|:---:|
| 0.10 | 1.645 | $2.00 > 1.645$ | Reject $H_0$ |
| 0.05 | 1.96 | $2.00 > 1.96$ | Reject $H_0$ |
| 0.01 | 2.576 | $2.00 < 2.576$ | Fail to reject $H_0$ |


Computing the 95% confidence interval:
$$
\bar{X} \pm 1.96 \cdot \widehat{SE} = 52 \pm 1.96(1) = [50.04, 53.96]
$$

Interpreting this, the 95% CI does not contain 50, which matches the rejection at $\alpha=.05$, so we have statistical significance saying that there is a difference.

# Problem 7

In './data/ames_prices.csv', create an indicator that equals 1 when Central.Air == 'Y' and 0 otherwise.

```python
# imports
import numpy as np
import pandas as pd

# loading in the data
df = pd.read_csv('./data/ames_prices.csv')

# making sure this variable is either Y or N
print(df['Central.Air'].head())

# making the indicator variable
df['central_air'] = (df['Central.Air'] == 'Y').astype(int)

# checking that this worked
print(df.head())
```

a. Estimate the proportion of homes with central air.

```python
# printing the counts for each value
df['central_air'].value_counts()

# counting the number of homes with central air
has_central_air = (df['central_air'] == 1).sum()
# counting the number of homes without central air
has_no_central_air = (df['central_air'] == 0).sum()

# computing the proportionf
proportion = has_central_air / (has_no_central_air + has_central_air)
print(proportion)
```

> So about 93.3% of homes have central air.

b. Using a reproducible random-number seed, generate at least 2,000 bootstrap estimates of this proportion. Report the seed and number of bootstrap replications.

```python
# setting the random seed and num of bootstraps
random_seed = 5030
rng = np.random.default_rng(random_seed)
num_bootstraps = 5000

# generating the boostrap samples
# initializing empty list to store bootstrap samples
bootstrap_samples = []

# looping over each bootstrap
for boot in range(num_bootstraps):
    # creating the sample
    sample = rng.choice(df['central_air'].to_numpy(), size=len(df), replace=True)
    # computing the sample mean
    sample_mean = sample.mean()
    # appending the sample mean
    bootstrap_samples.append(sample_mean)

# converting to an array to work with these sample means later
bootstrap_samples = np.array(bootstrap_samples)

# printing the seed and num of bootstraps
print(f"The random seed is:", random_seed)
print(f"The number of boostrap replications is:", num_bootstraps)
print(f"The mean of the sample means from bootstrapping is:", bootstrap_samples.mean())
print(f"The difference between the bootstrap mean and the proportion in the data is", proportion - bootstrap_samples.mean())
```

c. Compute the bootstrap standard error as the sample standard deviation of the bootstrap estimates (using ddof=1 ensures that there's the 1/(N−1) degrees of freedom correction).

```python
# using the build in numpy function to get the SE/sample SD 
bootstrapping_se = np.std(bootstrap_samples, ddof=1)

print(f"The bootstrapping standard error is:", np.round(bootstrapping_se,5))
```

d. Using the original estimate and the bootstrap standard error, construct Normal 90%, 95%, and 99% confidence intervals.

```python
# creating the confidence intervals with the original estimate and the SE of 0.00463

# 90% confidence interval
ci_90 = (proportion - 1.645 * bootstrapping_se, proportion + 1.645 * bootstrapping_se)
print(f"The 90% CI is:", np.round(ci_90[0],5), np.round(ci_90[1],5))

# 95% confidence interval
ci_95 = (proportion - 1.96 * bootstrapping_se, proportion + 1.96 * bootstrapping_se)
print(f"The 95% CI is:", np.round(ci_95[0],5), np.round(ci_95[1],5))

# 99% confidence interval
ci_99 = (proportion - 2.576 * bootstrapping_se, proportion + 2.576 * bootstrapping_se)
print(f"The 99% CI is:", np.round(ci_99[0],5), np.round(ci_99[1],5))
```

e. Use each confidence interval to evaluate H0:p=.90. At the matching significance levels α=.10,.05,.01, state whether you reject or fail to reject the null hypothesis. Briefly interpret your results.

> Looking at these confidence intervals, we can see that 0.90 does not appear within any of them. Because 0.90 is not included, we can reject the null hypothesis that p=0.90 at all three of these signficance levels. The confidence intervals gives us the range of where we think the proportions will lie if we repeated this simulation with the same assumptions and data, and since 0.9 is not included in that range all the way down till the 90% confidence level, we are pretty sure that the true proportion of homes with central air is not 0.90.

# Problem 8

In ./data/ames_prices.csv, estimate the Pearson correlation between np.arcsinh(area) and np.arcsinh(price).

```python
# loading in the data
ames = pd.read_csv('./data/ames_prices.csv')
#print(ames.head())

# creating new columns with the applied arcsinh function to the columns
ames['asinh_area'] = np.arcsinh(ames['area'])
ames['asinh_price'] = np.arcsinh(ames['price'])

# computing the Pearson correlation
correlation_8a = np.corrcoef(ames['asinh_area'], ames['asinh_price'])[0, 1]
print(f"The Pearson correlation between arcsinh(area) and arcsinh(price) is:", np.round(correlation_8a,5))
```

a. Using a reproducible random-number seed, generate at least 2,000 bootstrap estimates of the correlation by resampling paired rows. Report the seed and number of bootstrap replications.

```python
# setting the random seed and num of bootstraps
random_seed = 5030
rng = np.random.default_rng(random_seed)
num_bootstraps = 5000
n = len(ames)

# pulling the columns to pair and converting them into into arrays to work with
area_arr = ames['asinh_area'].to_numpy()
price_arr = ames['asinh_price'].to_numpy()

# generating the bootstrap samples
# initializing empty list to store bootstrap samples of correlation
bootstrap_samples = []

# looping over each bootstrap
for boot in range(num_bootstraps):
    # rather than pulling random data points, we have to pull a random index and pull the values for the columns at that index
    idx = rng.integers(0, n, size=n)
    # computing the correlation on the boostrapped pairs
    sample_correlation = np.corrcoef(area_arr[idx], price_arr[idx])[0, 1]
    # appending the sample correlation to the list
    bootstrap_samples.append(sample_correlation)

# converting sample correlations to an array for processing
bootstrap_samples = np.array(bootstrap_samples)

# printing the seed and num of bootstraps
print(f"The random seed is: {random_seed}")
print(f"The number of bootstrap replications is: {num_bootstraps}")
print(f"The mean of the bootstrap correlations is: {bootstrap_samples.mean():.4f}")
print(f"The difference between the sample correlation and the bootstrap mean is: {correlation_8a - bootstrap_samples.mean():.6f}")
```

b. Compute the bootstrap standard error as the sample standard deviation of the bootstrap estimates (using ddof=1 ensures that there's the 1/(N−1) degrees of freedom correction).

```python
# using the build in numpy function to get the SE/sample SD 
bootstrapping_se = np.std(bootstrap_samples, ddof=1)

print(f"The bootstrapping standard error is:", np.round(bootstrapping_se,5))
```

c. Using the original estimate and the bootstrap standard error, construct centered-Normal 90%, 95%, and 99% confidence intervals.

```python
# creating the confidence intervals

# 90% confidence interval
ci_90 = (correlation_8a - 1.645 * bootstrapping_se, correlation_8a + 1.645 * bootstrapping_se)
print(f"The 90% CI is:", np.round(ci_90[0],5), np.round(ci_90[1],5))

# 95% confidence interval
ci_95 = (correlation_8a - 1.96 * bootstrapping_se, correlation_8a + 1.96 * bootstrapping_se)
print(f"The 95% CI is:", np.round(ci_95[0],5), np.round(ci_95[1],5))

# 99% confidence interval
ci_99 = (correlation_8a - 2.576 * bootstrapping_se, correlation_8a + 2.576 * bootstrapping_se)
print(f"The 99% CI is:", np.round(ci_99[0],5), np.round(ci_99[1],5))
```

> Since ρ=0 is not included in any of these confidence intervals, we can reject the null hypothesis H0:ρ=0 at all three significance levels. This is because all of the intervals spread over values much larger than 0, tells us that homes with larger living areas are positively correlated with higher sales prices (correlation does not equal causation, and here we have not controlled for other effects like area or house age that would allow us to isolate the casuality).

e. Briefly explain why area and price must remain paired during resampling.

> The estimator we are looking at over our bootstrapping samples is correlation between the mutated price and size of homes. Since we are looking for correlation between two variables, the values of those two variables must stay paired during our resampling. Otherwise, we would randomly pick a price and randomly pick a size, and then computing the correlation between them will not mean anything. Correlation at its core is looking at how two varaibles move together, so keeping the values paired/tied to one house allow us to capture the relationship between the two varaibles rather than randomly picking values that belong to different houses and trying to understand their relationship. For example, if we did not pair our samples we could have one pair be [price of a large house, size of a small house] and t[price of a small hosue, size of a large house] and comparing these two would mean nothing for understanding the relation ship between size and price.

```python
# loading in the data
metabric = pd.read_csv('./data/metabric.csv')
#print(metabric.head())

# dropping missing values
survival = metabric['Overall Survival (Months)'].dropna()

# converting to an array
survival = metabric['Overall Survival (Months)'].to_numpy()

# defining length/num of observations
n = len(survival)
print(f"There are {n} observations.")

# computing the sample mean for all the data
mean_survival = survival.mean()
print(f"The mean survial time time in months is:", np.round(mean_survival,5))
```
